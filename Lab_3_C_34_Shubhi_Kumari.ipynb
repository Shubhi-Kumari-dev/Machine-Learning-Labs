## **Find-S Algorithm**

**The Find-S algorithm finds the most specific hypothesis that is consistent with all positive training examples. It starts with the most specific hypothesis and generalizes it whenever a positive example contains a different attribute value. Negative examples are ignored by the Find-S algorithm.**

```python
X=df.iloc[:,:-1].values
y=df.iloc[:,-1].values
```

```python
def find_s(X,y):
    hypothesis=None
    print("Initial Hypothesis: ",hypothesis)
    for i in range(len(X)):
        if y[i]=="Yes":
            if hypothesis is None:
                hypothesis=X[i].copy()
            else:
                for j in range(len(hypothesis)):
                    if hypothesis[j]!=X[i][j]:
                        hypothesis[j]="?"
        print(f"\nAfter Training Example {i+1}")
        print(hypothesis)
    return hypothesis
```

```python
final_hypothesis=find_s(X,y)
print("\nFinal Hypothesis")
print(final_hypothesis)
```


## **Candidate Elimination Algorithm**

**The Candidate Elimination algorithm is a concept learning algorithm used to find all hypotheses that are consistent with the given training examples.**

```python
# Function to implement Candidate Elimination Algorithm
def candidate_elimination(concepts, target):

    # Initialize Specific hypothesis with the first positive training example
    first_pos_index = target.index("Yes")
    specific = concepts[first_pos_index].copy()

    # Initialize General hypothesis with '?' for every attribute
    general = [["?" for _ in range(len(specific))]]

    # Display initial hypotheses
    print("\nInitial S:", specific)
    print("Initial G:", general)

    # Process each training example
    for i, h in enumerate(concepts):

        # ------------------------------------------------
        # Positive Training Example
        # ------------------------------------------------
        if target[i] == "Yes":

            # Generalize the Specific hypothesis
            for x in range(len(specific)):

                # If the attribute value is different,
                # replace it with '?'
                if h[x] != specific[x]:
                    specific[x] = "?"

            # Remove general hypotheses that do not
            # cover the positive example
            general = [
                g
                for g in general
                if all(
                    g[j] == "?" or g[j] == h[j] for j in range(len(specific))
                )
            ]

        # ------------------------------------------------
        # Negative Training Example
        # ------------------------------------------------
        else:

            new_general = []

            # Specialize the General hypotheses
            for g in general:

                # Check whether the hypothesis covers
                # the negative example
                covers_negative = all(
                    g[j] == "?" or g[j] == h[j] for j in range(len(specific))
                )

                # If it covers the negative example,
                # specialize it
                if covers_negative:

                    for x in range(len(specific)):

                        # Only specialize '?' positions
                        if g[x] == "?":

                            # Use the Specific boundary value
                            if specific[x] != "?" and specific[x] != h[x]:
                                # Create a copy of the hypothesis
                                hypothesis = g.copy()
                                hypothesis[x] = specific[x]

                                # Add if it is not already present
                                if hypothesis not in new_general:
                                    new_general.append(hypothesis)

                else:
                    # Keep hypotheses that do not cover
                    # the negative example
                    if g not in new_general:
                        new_general.append(g)

            # Update General boundary
            general = new_general

        # Remove duplicate hypotheses
        unique_general = []

        for g in general:
            if g not in unique_general:
                unique_general.append(g)

        general = unique_general

        # Display the boundaries after each example
        print(f"\nExample {i + 1} ({'Positive' if target[i] == 'Yes' else 'Negative'}):")
        print("Instance:", h)
        print("S =", specific)
        print("G =", general)

    # Return final boundaries
    return specific, general


# ------------------------------------------------
# Load CSV and Run Algorithm
# ------------------------------------------------
df = pd.read_csv("enjoy.csv")

# Extract features (concepts) and target labels
X = df.iloc[:, :-1].values.tolist()
y = df.iloc[:, -1].values.tolist()

# Run Candidate Elimination
S, G = candidate_elimination(X, y)

# Display final boundaries
print("\n" + "=" * 40)
print("Final Specific Boundary:")
print(S)

print("\nFinal General Boundary:")
for g in G:
    print(g)
print("=" * 40)
```
